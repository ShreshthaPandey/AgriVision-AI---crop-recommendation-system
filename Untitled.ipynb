```python


price.columns = ['Group', 'Commodity', 'MSP', 'Price_15', 'Price_14', 'Price_13',
                 'Arr_15', 'Arr_14', 'Arr_13']

price = price.dropna(subset=['Commodity'])
price['Commodity'] = price['Commodity'].astype(str).str.strip()

print(price.shape)
print(price[['Group', 'Commodity', 'MSP', 'Price_15']].to_string())
```

```python
# Labour aur irrigation ko 0-1 scale par lao, phir mila do
crop['Effort_score'] = (
    0.7 * crop['Labour_days'] / crop['Labour_days'].max() +
    0.3 * crop['Irrigations'] / crop['Irrigations'].max()
)

# Low / Medium / High (3 barabar hisson mein)
crop['Effort'] = pd.qcut(crop['Effort_score'], 3, labels=['Low', 'Medium', 'High'], duplicates='drop')

print(crop.groupby('Crop')[['Effort_score', 'Effort']].first())
```

```python
def effort_label(x):
    if pd.isna(x):
        return None
    if x < 0.15:
        return 'Low'
    elif x < 0.35:
        return 'Medium'
    else:
        return 'High'

crop['Effort'] = crop['Effort_score'].apply(effort_label)

print(crop.groupby('Crop')['Effort'].first().value_counts())
```

```python
import pandas as pd, numpy as np, requests, time

districts = pd.read_csv("district_latlon.csv")

fix_names = {
    'Visakhapatanam': 'Visakhapatnam',
    'Spsr Nellore': 'Nellore',
    'East Godavari': 'Kakinada',
    'Prakasam': 'Ongole',
    'Leparada': 'Basar',
    'Lower Subansiri': 'Ziro',
    'East Siang': 'Pasighat',
    'West Siang': 'Aalo',
    'Upper Siang': 'Yingkiong',
    'Papum Pare': 'Itanagar',
    'Dibang Valley': 'Anini',
    'East Kameng': 'Seppa',
    'Upper Subansiri': 'Daporijo',
    'South Andamans': 'Port Blair',
    'Nicobars': 'Car Nicobar',
    'North And Middle Andaman': 'Mayabunder',
}

def geo(name, state):
    try:
        r = requests.get("https://geocoding-api.open-meteo.com/v1/search",
                         params={"name": name, "count": 10, "country_code": "IN"}, timeout=20)
        res = r.json().get("results", [])
        for x in res:
            if state.lower() in str(x.get("admin1", "")).lower():
                return x["latitude"], x["longitude"]
        return (res[0]["latitude"], res[0]["longitude"]) if res else (np.nan, np.nan)
    except Exception:
        return (np.nan, np.nan)

for i, row in districts[districts['lat'].isna()].iterrows():
    if row['District'] in fix_names:
        la, lo = geo(fix_names[row['District']], row['State'])
        districts.at[i, 'lat'] = la
        districts.at[i, 'lon'] = lo
        time.sleep(0.2)

print(districts['lat'].isna().sum(), "abhi bhi missing")
```

```python
state_avg = districts.groupby('State')[['lat', 'lon']].transform('mean')
districts['lat'] = districts['lat'].fillna(state_avg['lat'])
districts['lon'] = districts['lon'].fillna(state_avg['lon'])

print(districts['lat'].isna().sum(), "abhi bhi missing")
print(districts[districts['lat'].isna()]['State'].unique())
```

```python
# Calculate state-wise average coordinates
state_avg = districts.groupby('State')[['lat', 'lon']].transform('mean')

# Fill missing coordinates using state averages
districts['lat'] = districts['lat'].fillna(state_avg['lat'])
districts['lon'] = districts['lon'].fillna(state_avg['lon'])

# Check remaining missing values
print("Missing latitude:", districts['lat'].isna().sum())
print("Missing longitude:", districts['lon'].isna().sum())

# Show states whose latitude is still missing
print("States still missing:")
print(districts.loc[districts['lat'].isna(), 'State'].unique())
```

```python
# Approximate coordinates: latitude, longitude
manual = {
    'Delhi': (28.6139, 77.2090),       # New Delhi
    'Meghalaya': (25.5788, 91.8933)    # Shillong
}

for st, (la, lo) in manual.items():
    mask = (
        (districts['State'] == st) &
        (districts['lat'].isna() | districts['lon'].isna())
    )

    districts.loc[mask, 'lat'] = districts.loc[mask, 'lat'].fillna(la)
    districts.loc[mask, 'lon'] = districts.loc[mask, 'lon'].fillna(lo)

# Verify
print("Missing latitude:", districts['lat'].isna().sum())
print("Missing longitude:", districts['lon'].isna().sum())
```

```python
import requests
import pandas as pd
import numpy as np
import time

# District -> nearby city/town for coordinate lookup
fix_locations = {
    'Siang': ('Pasighat', 'Arunachal Pradesh'),
    'Anjaw': ('Hawai', 'Arunachal Pradesh'),
    'Tawang': ('Tawang', 'Arunachal Pradesh'),
    'Baksa': ('Mushalpur', 'Assam'),
    'Cachar': ('Silchar', 'Assam'),
    'Saran': ('Chhapra', 'Bihar'),
    'Ahmadabad': ('Ahmedabad', 'Gujarat'),
    'Dang': ('Ahwa', 'Gujarat'),
    'Narmada': ('Rajpipla', 'Gujarat'),
    'Tapi': ('Vyara', 'Gujarat'),
    'Mewat': ('Nuh', 'Haryana'),
    'Koderma': ('Koderma', 'Jharkhand'),
    'Raigad': ('Alibag', 'Maharashtra'),
    'Boudh': ('Boudh', 'Odisha'),
    'East District': ('Gangtok', 'Sikkim'),
    'North District': ('Mangan', 'Sikkim'),
    'South District': ('Namchi', 'Sikkim'),
    'West District': ('Geyzing', 'Sikkim'),
    'Gomati': ('Udaipur', 'Tripura'),
    'Allahabad': ('Prayagraj', 'Uttar Pradesh'),
    'Nadia': ('Krishnanagar', 'West Bengal')
}

def get_coordinates(place, state):
    try:
        response = requests.get(
            "https://geocoding-api.open-meteo.com/v1/search",
            params={
                "name": place,
                "count": 10,
                "country_code": "IN",
                "language": "en",
                "format": "json"
            },
            timeout=20
        )
        response.raise_for_status()

        results = response.json().get("results", [])

        for r in results:
            lat = r.get("latitude")
            lon = r.get("longitude")
            country = r.get("country_code", "").upper()
            api_state = str(r.get("admin1", "")).lower()

            if (
                country == "IN"
                and state.lower() in api_state
                and 6 <= lat <= 38
                and 68 <= lon <= 98
            ):
                return lat, lon

        return np.nan, np.nan

    except Exception as e:
        print(f"Error: {place}: {e}")
        return np.nan, np.nan


for district, (place, state) in fix_locations.items():
    mask = (
        districts['District'].eq(district)
        & districts['State'].eq(state)
    )

    if not mask.any():
        print("District not found:", state, district)
        continue

    lat, lon = get_coordinates(place, state)

    if pd.notna(lat) and pd.notna(lon):
        districts.loc[mask, 'lat'] = lat
        districts.loc[mask, 'lon'] = lon
        print(f"Fixed: {state} - {district}: {lat}, {lon}")
    else:
        print(f"Could not verify: {state} - {district}")

    time.sleep(0.3)
```
